```python
# 1. Define the columns in the exact 1-10 order again
ms_cols = [
    'ms_2_A1', 'ms_2_A2', 'ms_2_A3', 'ms_2_A4', 'ms_2_A5', # 1-5
    'ms_1_A1', 'ms_1_A2', 'ms_1_A3', 'ms_1_A4', 'ms_1_A5'  # 6-10
]

# 2. Get the Optimal value (The absolute minimum in the row)
df_merged['optimal_ms_val'] = df_merged[ms_cols].min(axis=1)

# 3. Get the value that Autochoice actually picked
# We use a lambda to look up the column based on the number in 'autochoice'
def get_autochoice_val(row):
    try:
        # Convert autochoice to int, subtract 1 for 0-based indexing
        choice_idx = int(row['autochoice']) - 1
        col_name = ms_cols[choice_idx]
        return row[col_name]
    except (ValueError, TypeError, IndexError):
        return np.nan

df_merged['autochoice_ms_val'] = df_merged.apply(get_autochoice_val, axis=1)

# 4. Calculate the "Regret Ratio"
# Ratio = (What we got) / (What we could have had)
df_merged['regret_ratio'] = df_merged['autochoice_ms_val'] / df_merged['optimal_ms_val']



# 5. Calculate "Percent Over Optimal" (The 'How Much Worse' metric)
df_merged['pct_worse'] = (df_merged['regret_ratio'] - 1) * 100
print(df_merged[df_merged['pct_worse']>30].sort_values(["wf_name", "inp_size"]).to_string())

# 6. Drop 'dur' columns to keep the output clean
df_merged = df_merged.drop(columns=df_merged.filter(like='dur').columns)

print("--- Accuracy vs. Quality ---")
accuracy = (df_merged['pct_worse'] == 0).mean() * 100
avg_loss = df_merged['pct_worse'].mean()

print(f"Strict Accuracy (0% loss): {accuracy:.2f}%")
print(f"Average Performance Loss: {avg_loss:.2f}%")

def error_severity(pct):
    if pct == 0: return "Optimal (0%)"
    if pct <= 1: return "Negligible (<1%)"
    if pct <= 5: return "Acceptable (1-5%)"
    return "Significant (>5%)"

df_merged['severity'] = df_merged['pct_worse'].apply(error_severity)
print("\n--- Error Distribution ---")
print(df_merged['severity'].value_counts(normalize=True) * 100)
```
